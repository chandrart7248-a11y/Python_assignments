```python
# Q1. Phone Contact Book
# --------------------------------------------------------------
# Create a dictionary to store contacts (name -> phone number).
# Show a small menu in a while loop:
#   1. Add contact
#   2. Search contact
#   3. Show all contacts
#   4. Exit

#   - While adding, if the name already exists, print
#     "Contact already exists".
#   - While searching, if the name is not found, print
#     "Contact not found".
```

```python
contacts = {}
while True:
  print('\n1. Add contact')
  print('2.search contact')
  print('3. Show all contacts')
  print('4. Exit')
  choose = input('Enter your choice: ')
  if choose == '1':
    name = input('Enter your contact name: ')
    if name in contacts:
      print('Contact already exists')
    else:
      phone = int(input('Enter phone number'))
      contacts[name] = phone
      print('Contact added')
  elif choose == '2':
    name = input('Enter contact name to search: ')
    if name in contacts:
      print('Phone number:', contacts[name])
    else:
      print('Contact not found')
  elif choose == '3':
    if len(contacts)==0:
      print('No contacts available')
    else:
      for name, phone in contacts.items():
        print(name, ':', phone)
  elif choose == '4':
    print('Exiting from contacts')
    break
  else:
    print('No such option')
```

```python
# Q2. Unique Visitors on a Website
# --------------------------------------------------------------
# A website logged visitor IDs throughout the day (with repeats):
#   visits = [101, 205, 101, 330, 205, 412, 101, 330, 509]

# Using a set, find and print:
#   - The number of total visits
#   - The number of unique visitors
#   - The unique visitor IDs in sorted order
```

```python
visits = [101, 205, 101, 330, 205, 412, 101, 330, 509]
unique_visits = set(visits)
print('No.of visitors: ', len(visits))
print('Unique visitors: ', len(unique_visits))
print('Unique visitor IDS in sorted: ', sorted(unique_visits))
```

```python

# Q3. Student Marks Dictionary
# --------------------------------------------------------------
# Marks of students are stored in a dictionary:
#   marks = {"Aman": 82, "Priya": 91, "Rahul": 45,
#            "Sneha": 67, "Kiran": 30}

# Write a program to:
#   - Print each student's name with "Pass" or "Fail"
#     (pass mark is 40)
#   - Find the topper and their marks
#   - Create a new dictionary with only the students who passed
```

```python
marks = {"Aman": 82, "Priya": 91, "Rahul": 45,
         "Sneha": 67, "Kiran": 30}
for name, score in marks.items():
    if score >= 40:
        print(name, "Pass")
    else:
        print(name, "Fail")
topper = ""
highest_marks = 0
for name, score in marks.items():
    if score > highest_marks:
        highest_marks = score
        topper = name
print("Topper:", topper)
print("Topper marks:", highest_marks)
passed_students = {}
for name, score in marks.items():
    if score >= 40:
        passed_students[name] = score
print("Passed students:", passed_students)
```

```python
# Q4. Common Friends
# --------------------------------------------------------------
# Two friends have these friend lists on a social app:
#   riya_friends  = {"Aman", "Neha", "Karan", "Sita", "Vikram"}
#   arjun_friends = {"Neha", "Vikram", "Pooja", "Aman", "Rohan"}

# Print:
#   - Mutual friends of Riya and Arjun
#   - Friends only Riya has
#   - All people in either list (without duplicates)
#   - Friend suggestions for Riya (Arjun's friends she doesn't know)
```

```python
riya_friends  = {"Aman", "Neha", "Karan", "Sita", "Vikram"}
arjun_friends = {"Neha", "Vikram", "Pooja", "Aman", "Rohan"}
mutual_friends = riya_friends & arjun_friends
print('Mutual friends: ', mutual_friends)
only_riya_friends = riya_friends - arjun_friends
print('Friends of only Riya: ', only_riya_friends)
all_people = riya_friends | arjun_friends
print('All People: ', all_people)
friends_suggestion_riya = arjun_friends - riya_friends
print('Friends suggestion for riya: ', friends_suggestion_riya)
```

```python
# Q5. Character Counter
# --------------------------------------------------------------
# Take a word or sentence as input. Using a dictionary, count how
# many times each character (ignoring spaces) appears. Then print
# the character that appears the most.

# Sample run:
#   Enter text: banana
#   {'b': 1, 'a': 3, 'n': 2}
#   Most frequent character: a (3 times)
```

```python
text = input('Enter your text: ')
char_count = {}
for char in text:
  if char == ' ':
    continue
  if char in char_count:
    char_count[char]= char_count[char]+1
  else:
    char_count[char]=1
print(char_count)
most_char= ''
highest_count = 0
for char, count in char_count.items():
  if count > highest_count:
    highest_count = count
    most_char = char
print('Most frequent character', most_char, '(', highest_count, 'times)')
```

```python
# Q6. Restaurant Menu Billing
# --------------------------------------------------------------
# A restaurant menu is stored as a dictionary:
#   menu = {"dosa": 60, "idli": 40, "vada": 30,
#           "coffee": 25, "tea": 15}

# Keep asking the customer what they want to order until they
# type "bill". For each item:
#   - If it is on the menu, add its price to the total
#   - If not, print "Item not available"
# At the end, print the total bill. If the total is more than
# Rs. 200, apply a 10% discount.
```

```python
menu = {"dosa": 60, "idli": 40, "vada": 30,
        "coffee": 25, "tea": 15}
total = 0
while True:
    item = input("Enter item (or type 'bill' to finish): ").lower()
    if item == "bill":
        break
    if item in menu:
        total = total + menu[item]
        print(f"Added {item} - Rs. {menu[item]}")
    else:
        print("Item not available")
print(f"\nTotal before discount: Rs. {total}")
if total >= 200:
    discount = total * 0.10
    total -= discount
    print(f"10% discount applied: Rs. {discount}")
print(f"Final Bill: Rs. {total}")
```

```python
# Q8. Library Book Tracker
# --------------------------------------------------------------
# A library stores book availability in a dictionary:
#   books = {"Python Basics": 3, "Data Science 101": 0,
#            "Machine Learning": 2, "SQL Guide": 1}
# (The number is how many copies are available.)

# Ask the user for a book name to borrow:
#   - If the book doesn't exist, print "Book not found"
#   - If copies are 0, print "Currently unavailable"
#   - Otherwise, reduce the count by 1 and print
#     "Book issued successfully"
# Print the updated dictionary at the end.
```

```python
books = {"Python Basics": 3, "Data Science 101": 0,
            "Machine Learning": 2, "SQL Guide": 1}
book_name = input('Enter the book name: ')
if book_name not in books:
  print('Book not found')
else:
  if books[book_name]==0:
    print('Currently unavailable')
  else:
    current_count = books[book_name]
    new_count = current_count - 1
    books[book_name]=new_count
    print('Books issued successfully')
print('updated book availability:')
print(books)
```

```python
# Q9. Group Employees by Department
# --------------------------------------------------------------
# Employee data is stored as a list of tuples (name, department):
#   employees = [("Ravi", "IT"), ("Anjali", "HR"),
#                ("Suresh", "IT"), ("Meena", "Finance"),
#                ("Arun", "HR"), ("Divya", "IT")]

# Create a dictionary where each department is a key and the
# value is a list of employee names in that department.
# Then print how many employees each department has.

# Expected dictionary:
#   {"IT": ["Ravi", "Suresh", "Divya"],
#    "HR": ["Anjali", "Arun"],
#    "Finance": ["Meena"]}
```

```python
employees = [
    ("Ravi", "IT"),
    ("Anjali", "HR"),
    ("Suresh", "IT"),
    ("Meena", "Finance"),
    ("Arun", "HR"),
    ("Divya", "IT")
]
departments = {}
for name, department in employees:
    if department in departments:
        departments[department].append(name)
    else:
        departments[department] = [name]
print(departments)
for department, names in departments.items():
    print(department, ":", len(names), "employees")
```

```python
# Q10. Quiz Answer Checker
# --------------------------------------------------------------
# The correct answers of a 5-question quiz are stored in a tuple:
#   answer_key = ("B", "C", "A", "D", "B")

# Ask the student to enter their 5 answers one by one. Compare
# each with the answer key and:
#   - Print which question numbers were correct and which wrong
#   - Store wrong question numbers in a set
#   - Print the final score out of 5 and a remark:
#       5     : Excellent
#       3 - 4 : Good
#       0 - 2 : Needs improvement
```

```python
answer_key = ("B", "C", "A", "D", "B")
score = 0
wrong_questions = set()
for i in range(5):
    answer = input("Enter answer for question " + str(i + 1) + ": ").upper()
    if answer == answer_key[i]:
        print("Question", i + 1, "is correct")
        score = score + 1
    else:
        print("Question", i + 1, "is wrong")
        wrong_questions.add(i + 1)
print("Wrong questions:", wrong_questions)
print("Final score:", score, "out of 5")
if score == 5:
    print("Excellent")
elif score >= 3:
    print("Good")
else:
    print("Needs improvement")
```
